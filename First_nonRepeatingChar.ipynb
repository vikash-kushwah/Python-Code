```python
"""
Given a string s consisting of lowercase Latin Letters. Return the first non-repeating character in s. If there is no non-repeating character, return '$'.
Examples:
Input: s = "geeksforgeeks"
Output: 'f'
Explanation: In the given string, 'f' is the first character in the string which does not repeat.
"""
```

```python
#Function to find the first non-repeating character in a string.
def nonRepeatingChar(s):                                     #time complexity: o(n**2)
        for i in s:
            if(s.count(i)==1):
                return i
        return '$'
print(nonRepeatingChar("geeksforgeeks"))
```

```python
def nonRepeatingChar(s):
    u, r = [], set()  # 'u' stores unique chars, 'r' stores repeating chars
    for c in s:
        if c in r:
            continue
        if c in u:
            u.remove(c)
            r.add(c)
        else:
            u.append(c)
    return u[0] if u else '$'

s="geeksforgeeks"
print(nonRepeatingChar(s))
print(nonRepeatingChar("hellomynameisvikash"))
```
