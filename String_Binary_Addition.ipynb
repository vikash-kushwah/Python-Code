```python
"""
Given two binary strings s1 and s2 consisting of only 0s and 1s. Find the resultant string after adding the two Binary Strings.
Note: The input strings may contain leading zeros but the output string should not have any leading zeros.

Input: s1 = "1101", s2 = "111"
Output: 10100
Explanation:
 1101
+ 111
10100
"""
```

```python
a="1011"
b="1001"
num1=int(a,2)
num2=int(b,2)
print(num1,num2)
res=num1+num2
print(bin(res)[2:])
```

```python
def addBinary(s1, s2):
		n1=int(s1,2)          
		n2=int(s2,2)
		res=n1+n2
		return bin(res)[2:]   #slicing res 
print(addBinary(a,b))
print(addBinary("100000","101"))
```
